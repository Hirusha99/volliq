```python
import cv2
import os

# Configuration: Define your classes and colors
CLASSES = ["Volleyball", "Player_A", "Judge", "Player_B"]
COLORS = [(0, 255, 0), (255, 0, 0), (0, 0, 255), (255, 255, 0)]

def draw_boxes(image_path, txt_path):
    img = cv2.imread(image_path)
    h, w, _ = img.shape
    
    with open(txt_path, 'r') as f:
        for line in f:
            parts = line.split()
            class_id = int(parts[0])
            x_c, y_c, box_w, box_h = map(float, parts[1:])
            
            # Convert normalized YOLO format to pixel coordinates
            x1 = int((x_c - box_w / 2) * w)
            y1 = int((y_c - box_h / 2) * h)
            x2 = int((x_c + box_w / 2) * w)
            y2 = int((y_c + box_h / 2) * h)
            
            # Draw box and label
            cv2.rectangle(img, (x1, y1), (x2, y2), COLORS[class_id], 2)
            cv2.putText(img, CLASSES[class_id], (x1, y1 - 10), 
                        cv2.FONT_HERSHEY_SIMPLEX, 0.5, COLORS[class_id], 2)
    
    return img

# Example usage
# img = draw_boxes("frame001.jpg", "frame001.txt")
# cv2.imshow("Annotation Visualization", img)
# cv2.waitKey(0)
```

```python


# Configuration
CLASSES = ["Volleyball", "Player_A", "Judge", "Player_B"]
COLORS = [(0, 255, 0), (255, 0, 0), (0, 0, 255), (255, 255, 0)]
OUTPUT_DIR = "annotated_frames_output"

# Create output folder if it doesn't exist
os.makedirs(OUTPUT_DIR, exist_ok=True)

def save_annotated_frame(image_path, txt_path, output_path):
    img = cv2.imread(image_path)
    if img is None:
        print(f"Failed to load: {image_path}")
        return
    
    h, w, _ = img.shape
    
    with open(txt_path, 'r') as f:
        for line in f:
            parts = line.split()
            if not parts: continue
            class_id = int(parts[0])
            x_c, y_c, box_w, box_h = map(float, parts[1:])
            
            # Convert normalized YOLO format to pixel coordinates
            x1 = int((x_c - box_w / 2) * w)
            y1 = int((y_c - box_h / 2) * h)
            x2 = int((x_c + box_w / 2) * w)
            y2 = int((y_c + box_h / 2) * h)
            
            # Draw box and label
            cv2.rectangle(img, (x1, y1), (x2, y2), COLORS[class_id], 2)
            cv2.putText(img, CLASSES[class_id], (x1, y1 - 10), 
                        cv2.FONT_HERSHEY_SIMPLEX, 0.5, COLORS[class_id], 2)
    
    # Save the file
    cv2.imwrite(output_path, img)

```

```python
# Example usage for a specific frame
image_file = "test_ann_data//AN01//frames//frame_0082.jpg"
label_file = "test_ann_data//AN01//labels//frame_0082.txt"
output_file = os.path.join(OUTPUT_DIR, "annotated_frame_0082.jpg")

save_annotated_frame(image_file, label_file, output_file)
print(f"Saved: {output_file}")
```
