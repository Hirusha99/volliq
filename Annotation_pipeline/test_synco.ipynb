```python
import librosa
import numpy as np

def get_audio_offset(ref_file, target_file):
    # 1. Load audio (load at low sr to speed up computation)
    y_ref, sr = librosa.load(ref_file, sr=16000)
    y_target, _ = librosa.load(target_file, sr=16000)
    
    # 2. Compute Cross-Correlation
    # We use correlate with 'full' mode to find the lag
    correlation = np.correlate(y_ref, y_target, mode='full')
    
    # 3. Find the index of the maximum correlation
    lags = np.arange(-len(y_target) + 1, len(y_ref))
    offset = lags[np.argmax(correlation)]
    
    # Convert samples to seconds
    offset_seconds = offset / sr
    return offset_seconds

```

```python
# List of your 4 synchronized files
files = ["E:/Hiru/data/24.08.2025/M3/First half/Final/A-E.mp4", "E:/Hiru/data/24.08.2025/M3/First half/Final/H-E.mp4", "E:/Hiru/data/24.08.2025/M3/First half/Final/R-E.mp4", "E:/Hiru/data/24.08.2025/M3/First half/Final/M-E.mp4"]

# Assume files[0] is your reference (Ground Truth)
for i in range(1, len(files)):
    delay = get_audio_offset(files[0], files[i])
    print(f"Delay between {files[0]} and {files[i]}: {delay:.4f} seconds")
```
